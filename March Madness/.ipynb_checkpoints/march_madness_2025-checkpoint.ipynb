```python
# import stats
import pandas as pd
espn_stats = pd.read_excel("OneDrive/Documents/GitHub/work_to_show/March Madness/Espn Stats.xlsx")
kenpom = pd.read_excel("OneDrive/Documents/GitHub/work_to_show/March Madness/kenpom-2025.xlsx")
# kenpom needs to read at line 0
kenpom.columns = kenpom.iloc[0]  # set first row as header
kenpom = kenpom[1:].reset_index(drop=True)  # remove that row from data
```

```python
kenpom.head()
```

```python
kenpom.dtypes        # dtype of every column
```

```python
kenpom = kenpom.apply(pd.to_numeric, errors='coerce')
```

```python
print(espn_stats.columns.tolist())
```

```python
# what is the most important feature? espn_stats
VALUES=['FG%','3P%']
TARGET_COL="PTS"
#Divide data into train and test sets
import numpy as np
np.random.seed(42)
espn_stats['r'] = np.random.uniform(size=len(espn_stats))
train=espn_stats[espn_stats["r"] <= .6]
test=espn_stats[espn_stats["r"] > .6]

#Random Forest in Python
from sklearn.metrics import confusion_matrix
X=train[VALUES]
y=train[TARGET_COL]
Xtest=test[VALUES]
ytest=test[TARGET_COL]
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(n_estimators=500, random_state=17)
rf.fit(X,y)

#Additional Benefit of Random Forest Models: Feature Selection
importances = rf.feature_importances_
sorted_indices = np.argsort(importances)[::-1]
feat_labels = X.columns[0:]
for f in range(X.shape[1]):
    print("%2d) %-*s %f" % (f + 1, 30,feat_labels[sorted_indices[f]],importances[sorted_indices[f]]))

#Feature Importance Graphed
from matplotlib import pyplot as plt
plt.title('Feature Importance')
plt.bar(range(X.shape[1]), importances[sorted_indices], align='center')
plt.xticks(range(X.shape[1]), X.columns[sorted_indices], rotation=90)
plt.tight_layout()
plt.show()
```
