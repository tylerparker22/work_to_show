```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(path_to_file)

df = pd.read_sql_query("SELECT * FROM enrollment_master;", conn)
conn.close()

df.head()

df = df.replace('None', pd.NA)
df = df[df['t_number'].notna()]

df.head()
```

```python
pivot_table = df.groupby(['academic_period', 'major_desc', 'ethnicity_code'])['t_number'].count().reset_index()
pivot_table.columns = ['academic_period', 'major_desc', 'ethnicity_code', 'count']
pivot_table
```

```python
import matplotlib.pyplot as plt

eth_year_major = df.groupby(['academic_period', 'ethnicity_code', 'major_desc'])['t_number'].count().reset_index()
eth_year_major.columns = ['academic_period', 'ethnicity_code', 'major_desc', 'count']

for period, group in eth_year_major.groupby('academic_period'):
    pivot = group.pivot_table(index='major_desc', columns='ethnicity_code', values='count', fill_value=0)
    
    if pivot.empty:
        continue
    
    fig, ax = plt.subplots(figsize=(16, 8))
    pivot.plot(kind='bar', ax=ax)
    ax.set_title(f'Academic Period: {period} — Students by Major & Ethnicity')
    ax.set_xlabel('Major')
    ax.set_ylabel('Count of Students')
    ax.legend(title='Ethnicity Code', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.xticks(rotation=60, ha='right')
    plt.tight_layout()
    plt.show()
```
