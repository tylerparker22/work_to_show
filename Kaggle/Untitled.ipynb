```python
import pandas as pd
df = pd.read_csv("C:/Users/tyler/OneDrive/Documents/GitHub/work_to_show/Kaggle/spotify_global_trends.csv")
df.head()
```

```python
%%R -i df
colnames(df) # col names 
unique(df$genre) # unique items in the genre
```

```python
# genre streamed the most
genre_streamed = (
    df
    .groupby('genre')                         # pandas uses groupby, not group_by
    .agg(streams=('streams', 'sum'))          # sum streams per genre
    .sort_values('streams', ascending=False)  # sort descending
    .reset_index()                            # bring genre back as a column
)
genre_streamed
```
