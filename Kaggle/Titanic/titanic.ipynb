```python
import pandas as pd
test = pd.read_csv('test.csv', sep = ',')
train = pd.read_csv('train.csv', sep = ',')
```

```python
# test.head()
# print(test.columns.tolist()) # ['PassengerId', 'Pclass', 'Name', 'Sex', 'Age', 'SibSp', 'Parch', 'Ticket', 'Fare', 'Cabin', 'Embarked']
#
train.head()
print(train.columns.tolist()) # ['PassengerId', 'Survived', 'Pclass', 'Name', 'Sex', 'Age', 'SibSp', 'Parch', 'Ticket', 'Fare', 'Cabin', 'Embarked']
```

```python
var1 = ['PassengerId', 'Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Ticket']
var2 = 'Survived'
```

```python
#split to training and test datasets
import numpy as np
np.random.seed(42)
train['r'] = np.random.uniform(size=len(train))
#train.head()
train=train[train["r"] <= .6]
#train.head()
test=train[train["r"] > .6]
#test.head()
train=train.drop("r",axis=1)
test=test.drop("r",axis=1)
```

```python
#regression on training data
import statsmodels.api as sm
import matplotlib.pyplot as plt
y=train[var2]
x=train[var1]
x = sm.add_constant(x) #needed to force an intercept into the model
reg = sm.OLS(y,x).fit()
print(reg.summary())
```

```python
#mean of absolute error (MAE) for test data
tx=test[var1]
tx=sm.add_constant(tx)
psales=reg.predict(tx)
tsales=test[var2]
ae=abs(tsales-psales)
mae=np.mean(ae)
print("The mean absolute error is",mae)
```
