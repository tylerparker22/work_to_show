```r
survey <- read.csv("C:/Users/tyler/OneDrive/Documents/GitHub/work_to_show/Classes Spring 2026/BDA Spring 26/Datasets/survey.csv")
head(survey)
colnames(survey)
```

```r
# install.packages("arules") #run only once
library("arules")

#identify transactions
transactions <- as(split(survey$Answer, survey$Stdnt_Nbr),"transactions")

# use the apriori algorithm for association analysis
# specify the minimum support and minimum confidence
# maxlen=2 tells R to only consider pairwise associations
rules<-apriori(transactions,parameter=list(support=0.02,confidence=0.1,maxlen=2))

#include the odds ratio, use the Haldane-Anscombe correction, sort by odds ratio
ci <- confint(rules, "oddsRatio", smoothCounts = .5)

quality(rules) <- cbind(
quality(rules),
oddsRatio = interestMeasure(rules, "oddsRatio", smoothCounts = .5),
oddsRatio = ci)

rules <- sort(rules, by = "oddsRatio")
inspect(rules)
```

```r
df <- read.csv("C:/Users/tyler/OneDrive/Documents/GitHub/work_to_show/Classes Spring 2026/BDA Spring 26/Datasets/mb.csv")
colnames(df)
head(df)
```

```r
#install.packages("arules") #run only once
library("arules")
#identify transactions
# transactions <- as(split(df$y, df$x),"transactions")
transactions <- as(split(df$itemtype, df$visitid),"transactions")

#use the apriori algorithm for association analysis
#specify the minimum support and minimum confidence
#maxlen=2 tells R to only consider pairwise associations
rules<-apriori(transactions,
               parameter=
                 list(support=0.02,confidence=0.1,maxlen=2)
               )

#include the odds ratio, use the Haldane-Anscombe correction, sort by odds ratio
ci <- confint(rules, "oddsRatio", smoothCounts = .5)
quality(rules) <- cbind(
  quality(rules),
  oddsRatio = interestMeasure(rules, "oddsRatio", smoothCounts = .5),
  oddsRatio = ci)

rules <- sort(rules, by = "oddsRatio")
inspect(rules)

# Convert rules to a data frame and export to CSV
rules_df <- as(rules, "data.frame")

# Also add LHS and RHS as separate columns for readability
rules_df$lhs <- labels(lhs(rules))
rules_df$rhs <- labels(rhs(rules))

write.csv(rules_df, "C:/Users/tyler/OneDrive/Documents/GitHub/work_to_show/Classes Spring 2026/BDA Spring 26/mb_rules.csv", row.names = FALSE)

cat("Exported", nrow(rules_df), "rules to CSV.\n")
```
