```python
import pandas as pd
from matplotlib.style.core import library

dbp = pd.read_csv('Datasets/dbp data for covariate.csv', sep = ',')
dbp
```

```python
# run this by itself to run r
%load_ext rpy2.ipython
```

```python
%%R -i dbp
dbp$Treatment <- as.factor(dbp$Treatment)
fit <- lm(DBP ~ Treatment + Weight, data = dbp)
anova(fit)

library(emmeans)
emm <- emmeans(fit, ~ Treatment)
pairs(emm)
emm
```
