```python
import pandas as pd
from matplotlib.style.core import library
from statsmodels.iolib.summary import summary

ac = pd.read_csv('Datasets/Automated checkouts.csv', sep = ',')
```

```python
# run this by itself to run r
%load_ext rpy2.ipython
```

```python
%%R -i ac
# Covariate
ac$Type <- as.factor(ac$Type)
fit <- lm(EndSales ~ Type + PredictedSales, data = ac)
anova(fit)
```

What was significant? The covariate of Predicted Sales was significant
H0: sales with automated checkouts = sales with traditional checkouts
H0: slope = 0 (predictsales has no impact on endsales) (rejected)
Follow-up? None, the primary independent variable of concern was not
significant
Conclusion: the impact of type of check-out could not be shown to impact
sales.


```python
import pandas as pd
line_down = pd.read_csv('Datasets/Line downtime.csv', sep = ',')
print(line_down.columns.tolist()) # ['day', 'line', 'method', 'downtime']
```

```python
%%R -i line_down
# block
library(agricolae)
fit <- aov(downtime ~ method + line, data = line_down)
print(summary(fit))
duncan_results <- duncan.test(fit, "method", group = TRUE)
print(duncan_results)
```

Method: Block Design
What was significant? Both the block (line) and method of maintenance were found to
significantly impact downtime
Follow-up? Duncan’s method of mean separation
Conclusion: The new method of maintenance improves down time:
Average Down Time
Old method of maintenance: 35.3a
New Method of maintenance: 30.7b

```python
import pandas as pd
pt = pd.read_csv('Datasets/Packaging trial.csv', sep = ',')
pt
```

```python
%%R -i pt
fit<-aov(duration~packaging,data=pt)
summary(fit)
#install.packages("agricolae") # if not installed
library(agricolae)
duncan_results <- duncan.test(fit, "method", group = TRUE)
print(duncan_results)
```

Method: 1-way ANOVA
What was significant? Packaging was found to significantly impact duration (p-value of 2.63e-05)
Follow-up? Duncan’s method of mean separation
Conclusion: Brands were shown to significantly impact time until a packing error. Brands 5 and 3 had a
significantly longer duration until error than Brands 1 and 4.

Brand3 33.79985 ab
Brand2 32.85824 bc
Brand1 30.63713 cd
Brand4 28.92457 d
abcd: Means with different letters were found to be significantly different at the 0.05 level

```python
import pandas as pd
pie = pd.read_csv('Datasets/Pie ingredients.csv', sep = ',')
```

```python
%%R -i pie
fit <- aov(rating ~ oil*butter, data = pie)
summary(fit)
#install.packages("agricolae") # if not installed
library(agricolae)
fit <- aov(rating ~ butter, data = pie)
duncan_results <- duncan.test(fit, "butter", group = TRUE)
print(duncan_results)
```

Method: 2X2 Factorial
What was significant? Butter was found to significantly impact rating (p-value of 0.00071); neither
oil nor the interaction better butter and oil was significant
Follow-up? Duncan’s method of mean separation for butter
Conclusion: The old butter had a higher rating than the new butter
Brand Average Rating
old 6.8 a
new 5.8 b
ab: Means with different letters were found to be significantly different at the 0.05 level

# For circle.csv, a large city replaced a large number of its traffic lights with traffic circles (round-abouts); they want to compare the number of accidents (column “accidents”) at the intersections before and after the change (column “type”); however, they also know accidents vary by month (column “Month”).

```python
%%R -i circle
# assignment
# block
# ['Month', 'type', 'accidents']
library(agricolae)
fit <- aov(accidents ~ type + Month, data = circle)
#print(summary(fit))
duncan_results <- duncan.test(fit,
                              "type", # change this to right of "~"
                              group = TRUE)
print(duncan_results)
```
