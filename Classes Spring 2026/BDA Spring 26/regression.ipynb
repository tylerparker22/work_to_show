```python
%%R
library(tidyverse)
house <- read.csv("C:/Users/tyler/OneDrive/Documents/GitHub/work_to_show/Classes Spring 2026/BDA Spring 26/Datasets/housing.csv")
colnames(house)
```

```python
%%R
head(house)
model <- lm(saleprice ~ sqft, data = house)
summary(model)
#below this line “extra” code to provide a visual plot
plot(house$sqft,house$saleprice)
abline(model)
pred_interval <- predict(model, newdata=house, interval="prediction",level = 0.95)
lines(house$sqft, pred_interval[,2])
lines(house$sqft, pred_interval[,3])
```
