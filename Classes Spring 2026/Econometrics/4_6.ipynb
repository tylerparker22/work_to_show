```r
test <- read.csv('https://raw.githubusercontent.com/blacktreeM/econ/refs/heads/main/walmart_test.csv')
head(test)
colnames(test)
```

```r
# create time columns 
test$year <- substring(test$date, 1, 4)
test$month <- substring(test$date, 6, 7)
test$day <- weekdays(as.Date(test$date))
```

```r
# Create the combined store_item ID column
test$ID <- paste(test$store_nbr, test$item_nbr, sep = "_")
```
