```r
library(tidyverse)
train <- read.csv("https://raw.githubusercontent.com/blacktreeM/econ/refs/heads/main/train_weekly_sales.csv")
test <- read.csv("https://raw.githubusercontent.com/blacktreeM/econ/refs/heads/main/test_weekly_sales.csv")
```

```r
colnames(train) # colnames of train: 'Date''ID''Weekly_Sales'
colnames(test) # colnames of test: 'Date''ID'
```

```r
# create time columns in train
train$year <- substring(train$Date, 1, 4)
train$month <- substring(train$Date, 6, 7)
train$day <- weekdays(as.Date(train$Date))
train$days    <- as.numeric(substring(train$Date, 9, 10))
train$quarter <- quarters(as.Date(train$Date))
```

```r
# create time columns in test
test$year <- substring(test$Date, 1, 4)
test$month <- substring(test$Date, 6, 7)
test$day <- weekdays(as.Date(test$Date))
test$days    <- as.numeric(substring(test$Date, 9, 10))
test$quarter <- quarters(as.Date(test$Date))
```

```r
# need this for submission
test$id = paste(test$store_nbr, test$item_nbr, test$date, sep = '_')
test$weekly_sales = ifelse(is.na(test$units), 0, test$units)
```
