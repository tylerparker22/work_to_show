```r
train <- read.csv('https://raw.githubusercontent.com/blacktreeM/econ/refs/heads/main/walmart_train.csv')
train$year <- substring(train$date, 1, 4)
train$month <- substring(train$date, 6, 7)
train$day <- weekdays(as.Date(train$date))
```

```r
# Create the combined store_item ID column
train$ID <- paste(train$store_nbr, train$item_nbr, sep = "_")
```

```r
# task 1
for (i in unique(train$ID)){
    item = subset(train, ID== i) 
model_1 = lm(units ~ year + month + day, item)
print(i)
print(model_1)
}
```

```r
train$days = as.numeric(substring(train$date, 9, 10))
train$quarter = quarters(as.Date(train$date))
```

```r
# assignment
for (i in unique(train$ID)){
    item = subset(train, ID== i) 
model_2 = lm(units ~ year * quarter + month + day + days + I(days^2), item)
print(i)
print(model_2)
}
```

```r
test <- read.csv('https://raw.githubusercontent.com/blacktreeM/econ/refs/heads/main/walmart_test.csv')
head(test)
colnames(test)
```

```r
# create time columns 
test$year <- substring(test$date, 1, 4)
test$month <- substring(test$date, 6, 7)
test$day <- weekdays(as.Date(test$date))
test$days    <- as.numeric(substring(test$date, 9, 10))
test$quarter <- quarters(as.Date(test$date))
```

```r
# Create the combined store_item ID column
test$ID <- paste(test$store_nbr, test$item_nbr, sep = "_")
```

```r
for(i in unique(train$ID)) { 
    train_sub <- subset(train , ID == i)
    test_sub <- subset(test , ID == i)
    my_model = lm(log(units + 1) ~ year * quarter + month + day + days + I(days ^2), train_sub)
    prediction = exp(predict(my_model, test_sub)) -1
    test$units[test$ID == i] = prediction
    }
```

```r
test$id = paste(test$store_nbr, test$item_nbr, test$date, sep = '_')
```

```r
test$units = ifelse(is.na(test$units), 0, test$units)
```
