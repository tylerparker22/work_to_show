```python
%%R
library(tidyverse)
train<-read_csv('https://raw.githubusercontent.com/blacktreeM/econ/refs/heads/main/train.csv')
```

```python
%%R
head(train)
tail(train)
nrow(train)
colnames(train)
summary(train)
```

```python
%%R
# linear model
lm(count ~ season, data = train)
```

```python
%%R
# with factor 
lm(formula = count ~ factor(weather), data = train)
```

```python
%%R
#Do not use factor() for numerical variables!
lm(formula = count ~ factor(temp), data = train)
```

```python
%%R
# assignment 2/16
lm(count ~ season + weather + temp + humidity + windspeed + holiday, data = train)
```
